```python

# ============================================================
# PRACTICAL 6: DATA AGGREGATION
# ANALYZING SALES PERFORMANCE BY REGION
# Dataset: customer_shopping_data.csv
# ============================================================


# 1. IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")


# 2. IMPORT THE DATASET

df = pd.read_csv("customer_shopping_data.csv")

print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

display(df.head())


# 3. EXPLORE THE DATASET

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())


# 4. IDENTIFY RELEVANT VARIABLES

# Shopping mall represents the grouping location
# Category represents the product category
# Quantity represents units sold
# Price represents unit price

print("\nRelevant Columns:")
print("Location:", "shopping_mall")
print("Product Category:", "category")
print("Quantity:", "quantity")
print("Price:", "price")


# 5. DATA CLEANING

# Remove duplicate records

print("\nDuplicate Records Before Removal:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate Records After Removal:")
print(df.duplicated().sum())

# Handle missing values

print("\nMissing Values Before Treatment:")
print(df.isnull().sum())

df = df.dropna(
    subset=["shopping_mall", "category", "quantity", "price"]
)

# Remove extra spaces from text columns

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

print("\nMissing Values After Treatment:")
print(df.isnull().sum().sum())


# 6. CREATE SALES AMOUNT COLUMN

# Sales Amount = Quantity × Price

df["sales_amount"] = df["quantity"] * df["price"]

print("\nSales Amount Calculated Successfully!")

display(
    df[
        ["shopping_mall", "category", "quantity",
         "price", "sales_amount"]
    ].head()
)


# 7. CONVERT DATE COLUMN

df["invoice_date"] = pd.to_datetime(
    df["invoice_date"],
    dayfirst=True,
    errors="coerce"
)

print("\nUpdated Date Column:")
print(df["invoice_date"].head())


# 8. GROUP SALES DATA BY REGION / SHOPPING MALL

mall_sales = (
    df.groupby("shopping_mall")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal Sales Amount by Shopping Mall:")
display(mall_sales)


# 9. VISUALIZE SALES DISTRIBUTION USING BAR CHART

plt.figure(figsize=(12, 6))

mall_sales.plot(
    kind="bar",
    color="skyblue"
)

plt.title("Total Sales by Shopping Mall")
plt.xlabel("Shopping Mall")
plt.ylabel("Total Sales Amount")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 10. VISUALIZE SALES DISTRIBUTION USING PIE CHART

plt.figure(figsize=(9, 9))

plt.pie(
    mall_sales,
    labels=mall_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Shopping Mall")
plt.tight_layout()
plt.show()


# 11. IDENTIFY TOP-PERFORMING REGIONS

top_regions = mall_sales.sort_values(
    ascending=False
).head(5)

print("\nTop 5 Performing Shopping Malls:")
display(top_regions)

print("\nBest Performing Shopping Mall:")
print(top_regions.index[0])

print("\nHighest Sales Amount:")
print(top_regions.iloc[0])


# 12. GROUP SALES BY SHOPPING MALL AND PRODUCT CATEGORY

mall_category_sales = (
    df.groupby(["shopping_mall", "category"])
    ["sales_amount"]
    .sum()
    .reset_index()
)

print("\nSales by Shopping Mall and Product Category:")
display(mall_category_sales.head(20))


# 13. CREATE PIVOT TABLE

pivot_data = df.pivot_table(
    index="shopping_mall",
    columns="category",
    values="sales_amount",
    aggfunc="sum",
    fill_value=0
)

print("\nPivot Table of Sales:")
display(pivot_data)


# 14. CREATE STACKED BAR CHART

pivot_data.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 7),
    colormap="tab20"
)

plt.title("Category-wise Sales by Shopping Mall")
plt.xlabel("Shopping Mall")
plt.ylabel("Total Sales Amount")

plt.xticks(rotation=45)
plt.legend(
    title="Product Category",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# 15. CREATE GROUPED BAR CHART

pivot_data.plot(
    kind="bar",
    stacked=False,
    figsize=(15, 7),
    colormap="tab20"
)

plt.title("Comparison of Product Category Sales")
plt.xlabel("Shopping Mall")
plt.ylabel("Total Sales Amount")

plt.xticks(rotation=45)
plt.legend(
    title="Product Category",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# 16. FIND TOP-PERFORMING PRODUCT CATEGORIES

category_sales = (
    df.groupby("category")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal Sales by Product Category:")
display(category_sales)

plt.figure(figsize=(12, 6))

category_sales.plot(
    kind="bar",
    color="orange"
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 17. ANALYZE QUANTITY SOLD BY SHOPPING MALL

quantity_sales = (
    df.groupby("shopping_mall")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal Quantity Sold by Shopping Mall:")
display(quantity_sales)

plt.figure(figsize=(12, 6))

quantity_sales.plot(
    kind="bar",
    color="green"
)

plt.title("Total Quantity Sold by Shopping Mall")
plt.xlabel("Shopping Mall")
plt.ylabel("Quantity Sold")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 18. MONTHLY SALES ANALYSIS

df["month"] = df["invoice_date"].dt.month

monthly_sales = (
    df.groupby("month")["sales_amount"]
    .sum()
)

print("\nMonthly Sales:")
display(monthly_sales)

plt.figure(figsize=(10, 5))

monthly_sales.plot(
    kind="line",
    marker="o",
    color="red"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")

plt.grid(True)
plt.tight_layout()
plt.show()


# 19. FINAL SUMMARY

print("\n========== SALES PERFORMANCE SUMMARY ==========")

print("\nTotal Sales:")
print(df["sales_amount"].sum())

print("\nTotal Transactions:")
print(df["invoice_no"].nunique())

print("\nNumber of Shopping Malls:")
print(df["shopping_mall"].nunique())

print("\nNumber of Product Categories:")
print(df["category"].nunique())

print("\nTop Performing Shopping Mall:")
print(mall_sales.idxmax())

print("\nHighest Sales Amount:")
print(mall_sales.max())

print("\nTop Selling Product Category:")
print(category_sales.idxmax())


# 20. EXPORT AGGREGATED DATA

mall_sales.to_csv(
    "Sales_By_Shopping_Mall.csv"
)

category_sales.to_csv(
    "Sales_By_Category.csv"
)

mall_category_sales.to_csv(
    "Sales_By_Mall_And_Category.csv",
    index=False
)

pivot_data.to_csv(
    "Sales_Pivot_Table.csv"
)

print("\nAll Aggregated Files Exported Successfully!")


# CONCLUSION

print("""
CONCLUSION:

In this practical, data aggregation was performed
on the customer shopping dataset.

The dataset was imported and explored to understand
its structure and contents.

Sales amount was calculated using quantity and price.
Sales data was grouped by shopping mall and product
category to calculate total sales.

Bar charts and pie charts were created to visualize
sales distribution.

Top-performing shopping malls and product categories
were identified.

Stacked and grouped bar charts were used to compare
sales across shopping malls and product categories.

Thus, data aggregation was successfully performed
to analyze retail sales performance.
""")
```
