```python

# Install required libraries
# pip install qiskit qiskit-aer matplotlib

# Step 1: Import required libraries
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Step 2: Create a quantum circuit
# 3 qubits and 2 classical bits
qc = QuantumCircuit(3, 2)

# Step 3: Prepare the state to be teleported
# Apply X gate on qubit 0
qc.x(0)

# Step 4: Create entanglement between qubit 1 and qubit 2
qc.h(1)
qc.cx(1, 2)

# Step 5: Perform Bell-state measurement operations
qc.cx(0, 1)
qc.h(0)

# Step 6: Measure qubit 0 and qubit 1
qc.measure(0, 0)
qc.measure(1, 1)

# Step 7: Apply conditional correction gates
# Apply X gate if classical bit 1 is 1
with qc.if_test((1, 1)):
    qc.x(2)

# Apply Z gate if classical bit 0 is 1
with qc.if_test((0, 1)):
    qc.z(2)

# Step 8: Display the quantum circuit
print("Quantum Teleportation Circuit:")
print(qc.draw())

# Step 9: Create quantum simulator
simulator = AerSimulator()

# Step 10: Transpile the circuit
compiled_circuit = transpile(qc, simulator)

# Step 11: Execute the circuit
result = simulator.run(
    compiled_circuit,
    shots=1024
).result()

# Step 12: Display measurement results
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

# Step 13: Display histogram
plot_histogram(counts)
plt.show()
```
