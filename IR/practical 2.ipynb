```python

# Import defaultdict to create an inverted index easily
from collections import defaultdict

# Step 1: Create sample documents
documents = {
    1: "Information retrieval is important",
    2: "Information retrieval uses indexing",
    3: "Indexing improves document searching",
    4: "Searching documents is easy"
}

# Step 2: Create an empty inverted index
inverted_index = defaultdict(list)

# Step 3: Build the inverted index
for doc_id, text in documents.items():

    # Convert document text into lowercase
    text = text.lower()

    # Split the document into individual words
    words = text.split()

    # Visit each word in the document
    for word in words:

        # Add document ID if it is not already present
        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)

# Step 4: Display the inverted index
print("Inverted Index:")
for word, doc_ids in inverted_index.items():
    print(word, ":", doc_ids)

# Step 5: Accept search query from the user
query = input("\nEnter search query: ").lower()

# Split the query into individual words
query_words = query.split()

# Step 6: Retrieve documents containing all query words
result = None

for word in query_words:

    # Get document IDs containing the current word
    doc_ids = set(inverted_index.get(word, []))

    # Find common documents for all query words
    if result is None:
        result = doc_ids
    else:
        result = result.intersection(doc_ids)

# Step 7: Display the retrieved documents
print("\nRetrieved Documents:")

if result:
    for doc_id in sorted(result):
        print("Document", doc_id, ":", documents[doc_id])
else:
    print("No matching documents found.")
```
