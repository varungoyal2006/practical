```python

# Import the required libraries from NLTK
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import wordpunct_tokenize
from nltk.stem import PorterStemmer

# Download the English stopwords dataset if it is not already available
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Accept input text from the user
text = input("Enter the text: ")

# Step 1: Convert the entire text into lowercase
text = text.lower()

# Step 2: Tokenize the text into individual words
words = wordpunct_tokenize(text)

# Remove punctuation and numbers, keeping only alphabetic words
words = [word for word in words if word.isalpha()]

print("\nOriginal Tokens:")
print(words)

# Step 3: Remove stopwords from the tokenized text
# Stopwords are common words like is, the, an, and, etc.
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:")
print(filtered_words)

# Step 4: Perform stemming using Porter Stemmer
# Stemming reduces words to their root or stem form
ps = PorterStemmer()

stemmed_words = [ps.stem(word) for word in filtered_words]

print("\nAfter Stemming:")
print(stemmed_words)
```
